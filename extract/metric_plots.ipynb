```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
metrics = pd.read_csv("results.csv", )
metrics.head()
```

```python
metrics_glw6 = pd.read_csv("results_GLW6.csv")
metrics_glw6.head()
```

```python
plt.hist(metrics["iptm"])
plt.xlabel("iPTM score")
plt.ylabel("Count")
plt.title("Histogram of iPTM scores on Predicted Receptors")
```

```python
plt.hist(metrics["ptm"])
plt.xlabel("PTM score")
plt.ylabel("Count")
plt.title("Histogram of PTM scores on Predicted Receptors")
```

```python
published_results = pd.read_csv("published_metrics.csv")
published_results.head()
```

```python
plt.hist(published_results["iPTM"], color="orange", alpha=0.5)
plt.hist(metrics["iptm"], color="blue", alpha=0.5)
plt.xlabel("iPTM score")
plt.ylabel("Count")
plt.title("Histogram of iPTM scores")
```

```python
plt.hist(published_results["PTM"], color="orange", alpha=0.5)
plt.hist(metrics["ptm"], color="blue", alpha=0.5)
plt.xlabel("PTM score")
plt.ylabel("Count")
plt.title("Histogram of PTM scores")
```

```python
sns.kdeplot(metrics["iptm"], color="blue", alpha=0.5, fill=True)
sns.kdeplot(published_results["iPTM"], color="orange", alpha=0.5, fill=True)
plt.xlabel("iPTM score")
plt.title("Histogram of iPTM scores")
```

```python
sns.kdeplot(metrics["ptm"], color="blue", alpha=0.5, fill=True)
sns.kdeplot(published_results["PTM"], color="orange", alpha=0.5, fill=True)
plt.xlabel("PTM score")
plt.title("Histogram of PTM scores")
```

```python
sns.kdeplot(metrics["iptm"], color="blue", alpha=0.5, fill=True)
sns.kdeplot(metrics_glw6["iptm"], color="orange", alpha=0.5, fill=True)
plt.xlabel("iPTM score")
plt.title("Histogram of iPTM scores")
```
