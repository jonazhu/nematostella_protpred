```python
import pandas as pd
import numpy as np
```

```python
df_glw1 = pd.read_csv("results.csv")
df_glw6 = pd.read_csv("results_GLW6.csv")
df_glw1.head()
```

```python
df_glw6.head()
```

```python
df_glw6["names"] = [n[:-5] for n in df_glw6["names"]]
df_glw6.head()
```

```python
df_merged = df_glw1.merge(df_glw6, left_on='names', right_on='names')
df_merged.head()
```

```python
df_merged["iptm_diff"] = df_merged["iptm_x"] - df_merged["iptm_y"]
df_merged["ptm_diff"] = df_merged["ptm_x"] - df_merged["ptm_y"]
```

```python
df_merged.head()
```

```python
import matplotlib.pyplot as plt
plt.hist(df_merged["iptm_diff"])
plt.xlabel("Difference (GLWa1 - GLWa6)")
plt.title("Difference in iPTM value between GLWa1 and GLWa6 Prediction")
```

```python
plt.hist(df_merged["ptm_diff"])
plt.xlabel("Difference (GLWa1 - GLWa6)")
plt.title("Difference in PTM value between GLWa1 and GLWa6 Prediction")
```

```python
df_merged_iptm_sorted = df_merged.sort_values(by=["iptm_diff", "ptm_diff"], ascending=False)
df_merged_iptm_sorted.to_csv("diffs_iptm.csv")
df_merged_iptm_sorted.head()
```
