```python
import json
import os
import pandas as pd
import numpy as np
```

```python
df = pd.read_table("../data/NV2g.20240221.geneNames.txt", index_col="id")
df = df.dropna(subset=["name"])
df.head()
```

```python
folds_dir = "/Users/jonathanzhu/Downloads/folds"
files = os.listdir(folds_dir)
files = [f for f in files if (".zip" not in f and ".DS_Store" not in f)]
files
```

```python
names = []
iptms = []
ptms = []
ranking_scores = []
fraction_disordered = []
has_clash = []
chain_iptm_1 = []
chain_iptm_2 = []
chain_ptm_1 = []
chain_ptm_2 = []
chain_pair_iptm_1a = []
chain_pair_iptm_1b = []
chain_pair_iptm_2a = []
chain_pair_iptm_2b = []
chain_pair_pae_1a = []
chain_pair_pae_1b = []
chain_pair_pae_2a = []
chain_pair_pae_2b = []


for f in files:
    current_jobs = os.listdir(folds_dir + "/" + f)
    current_jobs = [c for c in current_jobs if (".md" not in c and ".DS_Store" not in c)]

    for c in current_jobs:
        current_json = folds_dir + "/" + f + "/" + c + "/fold_" + c + "_summary_confidences_0.json"
        names.append(c)
        with open(current_json, 'r') as fr:
            data = json.load(fr)
        iptms.append(data["iptm"])
        ptms.append(data["ptm"])
        ranking_scores.append(data['ranking_score'])
        fraction_disordered.append(data['fraction_disordered'])
        has_clash.append(data['has_clash'])

        #assertions: need to make sure structure of these is same across all. 
        #they should be but just in case
        assert len(data['chain_iptm']) == 2
        assert len(data['chain_ptm']) == 2
        assert len(data['chain_pair_iptm']) == 2
        assert len(data['chain_pair_pae_min']) == 2
        assert len(data['chain_pair_iptm'][0]) == 2
        assert len(data['chain_pair_iptm'][1]) == 2
        assert len(data['chain_pair_pae_min'][0]) == 2
        assert len(data['chain_pair_pae_min'][1]) == 2

        chain_iptm_1.append(data['chain_iptm'][0])
        chain_iptm_2.append(data['chain_iptm'][1])
        chain_ptm_1.append(data['chain_ptm'][0])
        chain_ptm_2.append(data['chain_ptm'][1])
        chain_pair_iptm_1a.append(data['chain_pair_iptm'][0][0])
        chain_pair_iptm_1b.append(data['chain_pair_iptm'][0][1])
        chain_pair_iptm_2a.append(data['chain_pair_iptm'][1][0])
        chain_pair_iptm_2b.append(data['chain_pair_iptm'][1][1])
        chain_pair_pae_1a.append(data['chain_pair_pae_min'][0][0])
        chain_pair_pae_1b.append(data['chain_pair_pae_min'][0][1])
        chain_pair_pae_2a.append(data['chain_pair_pae_min'][1][0])
        chain_pair_pae_2b.append(data['chain_pair_pae_min'][1][1])

results = pd.DataFrame({'names': names,
                        'iptm': iptms,
                        'ptm': ptms,
                        'ranking_score': ranking_scores,
                        'fraction_disordered': fraction_disordered,
                        'has_clash': has_clash,
                        'chain_iptm_1': chain_iptm_1,
                        'chain_iptm_2': chain_iptm_2,
                        'chain_ptm_1': chain_ptm_1,
                        'chain_ptm_2': chain_ptm_2,
                        'chain_pair_iptm_1a': chain_pair_iptm_1a,
                        'chain_pair_iptm_1b': chain_pair_iptm_1b,
                        'chain_pair_iptm_2a': chain_pair_iptm_2a,
                        'chain_pair_iptm_2b': chain_pair_iptm_2b,
                        'chain_pair_pae_1a': chain_pair_pae_1a,
                        'chain_pair_pae_1b': chain_pair_pae_1b,
                        'chain_pair_pae_2a': chain_pair_pae_2a,
                        'chain_pair_pae_2b': chain_pair_pae_2b})
```
