```python
import pandas as pd
import numpy as np
from Bio import SeqIO
```

```python
df_nvec = pd.read_csv("matches_Nvec.csv")
df_jaNem = pd.read_csv("matches_jaNemVect.csv")
df_nvec.head()
```

```python
df_jaNem.head()
```

```python
df_nvec["equal"] = df_nvec["protein1"] == df_nvec["protein2"]
df_nvec_unequal = df_nvec[df_nvec["equal"] == False]
df_nvec_unequal
```

```python
prots_nvec = []
with open("../../data/proteins_v2.fa") as handle:
    for record in SeqIO.parse(handle, "fasta"):
        prots_nvec.append(record)

prots_nvec
```

```python
def FindSeqRecord(name, seqs):
    for s in range(len(seqs)):
        if name == seqs[s].description:
            return s
    return -1

for i in range(len(df_nvec_unequal)):
    p1 = df_nvec_unequal.iloc[i]["protein1"]
    s1 = FindSeqRecord(p1[1:], prots_nvec)
    if s1 == -1:
        continue

    p2 = df_nvec_unequal.iloc[i]["protein2"]
    s2 = FindSeqRecord(p2[1:], prots_nvec)
    if s2 == -1:
        continue

    if str(prots_nvec[s1].seq) == str(prots_nvec[s2].seq):
        print(p1, "protein sequence is equal to", p2, "protein sequence")
    else:
        print(p1, "and", p2, "protein sequences are not equal")
        print(str(prots_nvec[s1].seq))
        print(str(prots_nvec[s2].seq))
        print()
```

```python
df_jaNem["equal"] = df_jaNem["protein1"] == df_jaNem["protein2"]
df_jaNem_unequal = df_jaNem[df_jaNem["equal"] == False]
df_jaNem_unequal
```

```python
prots_jaNem = []
with open("../../data/protein.faa") as handle:
    for record in SeqIO.parse(handle, "fasta"):
        prots_jaNem.append(record)

prots_jaNem
```

```python
for i in range(len(df_jaNem_unequal)):
    p1 = df_jaNem_unequal.iloc[i]["protein1"]
    s1 = FindSeqRecord(p1[1:], prots_jaNem)
    if s1 == -1:
        continue

    p2 = df_jaNem_unequal.iloc[i]["protein2"]
    s2 = FindSeqRecord(p2[1:], prots_jaNem)
    if s2 == -1:
        continue

    if str(prots_jaNem[s1].seq) == str(prots_jaNem[s2].seq):
        print(p1, "protein sequence is equal to", p2, "protein sequence")
    else:
        print(p1, "and", p2, "protein sequences are not equal")
        print(str(prots_jaNem[s1].seq))
        print(str(prots_jaNem[s2].seq))
        print()
```
