```python
gtf_jaNemVect = pd.read_table("../data/genomic.gtf", names=["chr", "source", "feature", "start", "end", "score", "strand", "frame", "attribute"])
gtf_jaNemVect.head()
```

```python
genes = gtf_jaNemVect[gtf_jaNemVect["feature"] == "transcript"]
genes.head()
```

```python
genes.iloc[0]["attribute"].split(";")
```

```python
#take the attribute column and return each item split by semicolon, removing the heading and quotes
def split_attribute(attr):
    splits = attr.split(";")
    for s in splits:
        s_trim = s.trim()
        #TODO: use regex to parse this and extract all the information between quotes
        
    gene_id = splits[0][9:-1]
    transcript_id = splits[1][16:-1]
    db_xref = splits[2][10:-1]
    gbkey = splits[3][8:-1]
    gene = splits[4][7:-1]
    evidence = splits[5][17:-1]
    prod = splits[6][10:-1]
    biotype = splits[7][21:-1]

    return gene_id, transcript_id, db_xref, gbkey, gene, evidence, prod, biotype

split_attribute(genes.iloc[0]["attribute"])
```

```python
chrs = []
starts = []
ends = []
geneids = []
transcriptids = []
dbxrefs = []
gbkeys = []
genenames = []
evidences = []
prods = []
biotypes = []
for i in range(len(genes)):
    current_attr = genes.iloc[i]["attribute"]
    current_chr = genes.iloc[i]["chr"]
    current_start = genes.iloc[i]["start"]
    current_end = genes.iloc[i]["end"]
    gene_id, transcript_id, db_xref, gbkey, gene, evidence, prod, biotype = split_attribute(current_attr)

    chrs.append(current_chr)
    starts.append(current_start)
    ends.append(current_end)
    geneids.append(gene_id)
    transcriptids.append(transcript_id)
    dbxrefs.append(db_xref)
    gbkeys.append(gbkey)
    genenames.append(gene)
    evidences.append(evidence)
    prods.append(prod)
    biotypes.append(biotype)

geneNames = pd.DataFrame({"chr": chrs,
                          "start": starts,
                          "end": ends,
                          "gene_id": geneids,
                          "transcript_id": transcriptids,
                          "db_xref": dbxrefs,
                          "gene_name": genenames,
                          "evidence": evidences,
                          "product": prods,
                          "biotype": biotypes})

geneNames.head()
```
