```python
import json
import copy
from Bio import SeqIO

import pandas as pd
import numpy as np
```

```python
prot_seqs = {}
prot_descs = {}
ids = []
with open("../data/protein.faa") as handle:
    for record in SeqIO.parse(handle, "fasta"):
        ids.append(record.id)
        prot_seqs[record.id] = record.seq
        prot_descs[record.id] = record.description
```

```python
#edit this for the jobs you want to do
#specifically, the GLWamide of your choice goes in the first sequence. So far we've done:
#glw1: QAGAPGLW
#glw6: SPKPPGLW
#glw9: LYGIGLW

#yet to do:
#glw2: QAGPPGLW
#glw3: 
#glw4: SAGPPGLW
#glw5: VAGPPGLW
#glw7:
#glw8: ANAPGLW

base_json_dict = {
    "name": "",
    "modelSeeds": [],
    "sequences": [
        {
            "proteinChain": {
                "sequence": "LYGIGLW",
                "count": 1,
                "useStructureTemplate": True
            }
        },
        {
            "proteinChain": {
                "sequence": "",
                "count": 1,
                "useStructureTemplate": True
                }
            }
    ],
    "dialect": "alphafoldserver",
    "version": 3
}
```

```python
#edit the lines where the comments are to ensure consistent naming

current_list = []
num_jsons = 1
for i in range(len(ids)):
    current_id = ids[i]
    try:
        current_desc = prot_descs[current_id]
    except:
        continue
    #print(current_name)
    if ("G-protein coupled receptor" in current_desc) or ("ion channel" in current_desc):
        current_seq = str(prot_seqs[current_id])
        current_dict = copy.deepcopy(base_json_dict)
        current_dict["name"] = current_id[:-2] + "_GLW9" #EDIT JOB TITLE WITH CORRECT GLW NUMBER
        current_dict["sequences"][1]["proteinChain"]["sequence"] = current_seq
        current_list.append(current_dict)
    if len(current_list) == 30 or i == len(ids) - 1:
        filename = "jaNemVect_GLW9_jobs_" + str(num_jsons) + ".json" #EDIT FILE NAME WITH CORRECT GLW NUMBER
        with open(filename, "w") as fw:
            json.dump(current_list, fw, indent=4)
        num_jsons += 1
        current_list = []
```
